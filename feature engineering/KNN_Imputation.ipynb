handling missing data using knn imputation

```python
import pandas as pd
```

```python
df = pd.read_csv('IPLPlayerStat.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=3 , weights='distance')
df['wickets'] = imputer.fit_transform(df[['wickets']])
```

```python
df['wickets'].isnull().sum()
```

```python
df.head()
```

but data imputed is in decimals ( wickets are not decimals )

```python
import numpy as np

df['wickets'] = np.ceil(df['wickets']).astype(int)
df.head()
```
