# Feature scaling and normalization

# Z - Score normalization

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('IPL_Matches_tail_300.csv')
df.head()
```

```python
scaler = StandardScaler()

df['team1_runs_scaled'] = scaler.fit_transform(df[['team1_runs']])
print(df[['team1_runs' , 'team1_runs_scaled']].head(10))
print(" ")
df['team2_runs_scaled'] = scaler.fit_transform(df[['team2_runs']])
print(df[['team2_runs' , 'team2_runs_scaled']].head(10))
```

```python
df['team1_wickets_scaled'] = scaler.fit_transform(df[['team1_wickets']])
print(df[['team1_wickets' , 'team1_wickets_scaled']].head(10))

print(" ")

df['team2_wickets_scaled'] = scaler.fit_transform(df[['team2_wickets']])
print(df[['team2_wickets' , 'team2_wickets_scaled']].head(10))
```

```python
df[['team1_runs' , 'team1_runs_scaled']].describe().round(2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,5))

plt.subplot(1,2,1) # 1 row , 2 columns , 1st plot
sns.kdeplot(df['team1_runs'] , fill = True)

plt.subplot(1,2,2)
sns.kdeplot(df['team1_runs_scaled'] , fill = True)
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1) # 1 row , 2 columns , 1st plot
sns.kdeplot(df['team1_wickets'] , fill = True)

plt.subplot(1,2,2)
sns.kdeplot(df['team1_wickets_scaled'] , fill = True)
```

# Min max normalization

```python
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('IPL_Matches_tail_300.csv')
df.head()
```

```python
scaler = MinMaxScaler()
df['team1_runs_scaled_minmax'] = scaler.fit_transform(df[['team1_runs']])
df[['team1_runs' , 'team1_runs_scaled_minmax']].head(10)
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.scatterplot( df['team1_runs'])

plt.subplot(1,2,2)
sns.scatterplot(df['team1_runs_scaled_minmax'])
```

```python
df[['team1_runs' , 'team1_runs_scaled_minmax']].describe()
```

# Mean normalization

```python
df = pd.read_csv('IPL_Matches_tail_300.csv')
df.head()
```

```python
mean = df['team1_runs'].mean()
min = df['team1_runs'].min()
max = df['team1_runs'].max()

print(mean)
print(min)
print(max)
```

```python
df['team1_runs_mean_scaled'] = (
    (df['team1_runs'] - mean) / ( max - min )
)

df[['team1_runs','team1_runs_mean_scaled']].head(10)
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.scatterplot( df['team1_runs'])

plt.subplot(1,2,2)
sns.scatterplot(df['team1_runs_mean_scaled'])
```

# Max Absolute Normalization

```python
from sklearn.preprocessing import MaxAbsScaler

df = pd.read_csv('IPL_Matches_tail_300.csv')
df.head()
```

```python
scaler = MaxAbsScaler()

df['team1_runs_mabs_scaled'] = scaler.fit_transform(df[['team1_runs']])
df[['team1_runs' , 'team1_runs_mabs_scaled']].head(10)
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.scatterplot( df['team1_runs'])

plt.subplot(1,2,2)
sns.scatterplot(df['team1_runs_mabs_scaled'])
```
