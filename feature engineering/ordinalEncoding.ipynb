```python
import pandas as pd

df = pd.read_csv('IPLPlayerStat.csv')
df.head()
```

```python
def player_form(avg):
  if avg <= 10 :
    return "Poor"
  elif avg <= 15 :
    return "Average"
  elif avg <= 20 :
    return "Good"
  elif avg <= 25 :
    return "Excellent"
  else :
    return "Outstanding"

df['player_form'] = df['batting_avg'].apply(player_form)

df[['player' , 'batting_avg' , 'player_form']].head()
```

```python
df['player_form'].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[['Poor' , 'Average' , 'Good' , 'Excellent' , 'Outstanding']]
)

df['player_form_encoded'] = encoder.fit_transform(df[['player_form']]).astype(int)
df[['player_form' , 'player_form_encoded']].head(10)
```

```python
df[['player' , 'batting_avg' , 'player_form' , 'player_form_encoded']].head(10)
```

```python
df.to_csv("IPLPlayerStat.csv")
```
