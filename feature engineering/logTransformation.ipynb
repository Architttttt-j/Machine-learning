```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('IPLPlayerStat.csv')
df.head()
```

```python
df['runs'].describe()
```

```python
sns.kdeplot(
    df['runs'] ,
    fill = True
)

print("right skewed data \n")
```

```python
df['runs_log'] = np.log1p(df['runs'])
df['runs_log10'] = df['runs_log'] / np.log(10)

df[['runs' , 'runs_log' , 'runs_log10']].head(10)
```

```python
sns.kdeplot(
    df['runs_log10'] ,
    fill = True
)

print('Less skewed data \n')
```
