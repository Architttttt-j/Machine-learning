```python
import pandas as pd

df = pd.read_csv("IPL_Matches_tail_300.csv")
df.head()
```

```python
df['team1'].unique()
```

```python
columns_to_update = ['team1' , 'team2' , 'toss_winner' , 'winner']
old_name = 'Royal Challengers Bangalore'
new_name = 'Royal Challengers Bengaluru'

for col in columns_to_update :
  df[col] = df[col].replace(old_name , new_name)

df['team2'].unique()
```

```python
df.to_csv('IPL_Matches_tail_300.csv')
```

```python
print(df['toss_decision'].unique())
print(df['team1'].nunique())
print(df['team2'].nunique())
```

# Apply One hot encoding

```python
df_encoded = pd.get_dummies(
    df,
    columns = ['team1' , 'team2' , 'toss_decision'],
    dtype = int
)

pd.set_option('display.max_columns' , None)

df_encoded.head(5)
```

drop first ( this will reduce the size and wont change meaning as well )

```python
df_encoded2 = pd.get_dummies(
    df ,
    columns = ['team1' , 'team2' , 'toss_decision'],
    drop_first=True ,
    dtype = int
)

df_encoded2.head()
```
