# mean imputation

we use mean values for missing data

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv('IPL_Matches_tail_300.csv')
print(df.shape)

df.info()
```

```python
df.isnull().sum()
```

```python
mean_df = df.copy()
mean_value = mean_df['team1_runs'].mean()

print(mean_value)
mean_df['team1_runs'].isnull().sum()
```

```python
mean_df['team1_runs'] = mean_df['team1_runs'].fillna(mean_value)
mean_df['team1_runs'].isnull().sum()
```

# Median Imputation

```python
median_df = df.copy()
median_value = median_df['team2_runs'].median()

print(median_value)

median_df['team2_runs'].isnull().sum()
```

```python
median_df['team2_runs'] = median_df['team2_runs'].fillna(median_value)
median_df['team2_runs'].isnull().sum()
```

# using scikit learn

```python
mean_imp = SimpleImputer(strategy='mean')
median_imp = SimpleImputer(strategy='median')
mode_imp = SimpleImputer(strategy = 'most_frequent')

df['team1_runs'] = mean_imp.fit_transform(df[['team1_runs']])
df['team2_runs'] = median_imp.fit_transform(df[['team2_runs']])
df['player_of_match'] = mode_imp.fit_transform(df[['player_of_match']])[: , 0]

df.isnull().sum()
```

but what about winners , i mean we can use mode , but the results wont be that accurate , we have to use mode but with some tweaks

```python
winner_counts = df['winner'].value_counts()

null_indices = df[df['winner'].isnull()].index
print(null_indices)
```

```python
for idx in null_indices :
  team1 = df.loc[idx , 'team1']
  team2 = df.loc[idx , 'team2']

  team1_wins = winner_counts.get(team1 , 0)
  team2_wins = winner_counts.get(team2 , 0)

  if team1_wins >= team2_wins :
    df.loc[idx , 'winner'] = team1
  else :
    df.loc[idx , 'winner'] = team2

df['winner'].isnull().sum()
```

```python
print("Rows where 'winner' was imputed:")
display(df.loc[null_indices, ['team1', 'team2', 'winner']])
```
