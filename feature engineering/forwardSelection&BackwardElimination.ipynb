```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('IPL_Matches_tail_300.csv')
df.head()
```

```python
data = df[['team1_runs' , 'team2_runs' , 'team1_wickets' , 'team2_wickets' , 'win_by_wickets']]
data.head()
```

```python

# find the relationship between every feature and the target domain ,  in IRL we use this with ML algorithms , but rn i havent learned it yet , so here we go

correlation = data.corr()
correlation
```

```python
target_corr = correlation['team1_runs'].drop('team1_runs')
target_corr
```

# Forward selection

```python
# arranging features from highest correlation to the lowest
target_corr = target_corr.reindex(
    target_corr.sort_values(ascending = False).index
)

target_corr
```

```python
# forward selection

def forward_selection(correlation_series):

  best_Feature = correlation_series.idxmax()
  correlation_series = correlation_series.drop(best_Feature)

  return correlation_series , best_Feature
```

```python
selected = []
remaining = target_corr.copy()

remaining , feature = forward_selection(remaining)
selected.append(feature)
print("selected feature : " , selected)
print("\n")
print("remaining feature :")

remaining
```

# Backward elimination

```python
# arranging features from lowest correlation to the highest
target_corr = target_corr.reindex(
    target_corr.sort_values(ascending = True).index
)

target_corr
```

```python
def backward_elimination(correlation_series):
  least_feature = correlation_series.idxmin()
  correlation_series = correlation_series.drop(least_feature)

  return correlation_series , least_feature
```

```python
remaining = target_corr.copy()
removed = []

remaining , feature = backward_elimination(remaining)
removed.append(feature)
print("\n removed features : " , removed)

print("\n remaining features : ")
remaining
```
