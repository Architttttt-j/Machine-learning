```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv('IPL_Matches_tail_300.csv')
df.head()
```

```python
df['team1'].unique()
```

```python
df['winner'].value_counts()
```

```python

# number of missing winners

df['winner'].isnull().sum()
```

```python
df['winner'] = df['winner'].fillna('No result')
df['winner'].tail(10)
```

```python
encoder = LabelEncoder()

df['winner_encoded'] = encoder.fit_transform(df['winner'])
df[['winner' , 'winner_encoded']].tail(15)
```

```python
mapping = pd.DataFrame({
    'winner' : encoder.classes_,
    'encoded value' : range(len(encoder.classes_))
})


mapping
```
