```python
df = df.dropna()
```

poor player_from is too much , i think i should remove it

```python
df = df[df['player_form'] != 'Poor']
df.head(10)
```

```python
df['player_form'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
cols = ['runs','boundaries','balls_faced','batting_avg','batting_strike_rate']
x_train , x_test , y_train , y_test = train_test_split(df[cols],df['player_form_encoded'],test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
y_train.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

x_train[:5]
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=11)
knn.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = knn.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
scores = []

for i in range(1,21):
  knn = KNeighborsClassifier(n_neighbors=i)
  knn.fit(x_train,y_train)
  y_pred = knn.predict(x_test)
  scores.append(accuracy_score(y_test,y_pred))
```

```python
plt.plot(range(1,21),scores)
```

```python
print(scores)
```

2D visualization

```python
cols = ['batting_avg','boundaries']
x_train2d , x_test2d , y_train2d , y_test2d = train_test_split(df[cols],df['player_form_encoded'])
```

```python
scaler2d = StandardScaler()
x_train2d = scaler.fit_transform(x_train2d)
x_test2d = scaler.fit_transform(x_test2d)
```

```python
knn2d = KNeighborsClassifier(n_neighbors=7)
knn2d.fit(x_train2d,y_train2d)
```

```python
y_pred = knn2d.predict(x_test2d)
accuracy_score(y_test2d,y_pred)
```

```python
from mlxtend.plotting import plot_decision_regions

fig,ax = plt.subplots(figsize=(15,10))
plot_decision_regions(X = x_train2d, y=y_train2d.values, clf=knn2d, legend=2, ax=ax)
plt.xlabel('batting_avg')
plt.ylabel('boundaries')
plt.title('decision surface for knn with batting average and boundaries')
plt.show()

print("\n"*2)
print("1 : Average , 2 : Good , 3 : Excellent and 4 : outstanding")
print("\n"*2)
```
