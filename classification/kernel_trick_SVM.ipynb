```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from matplotlib.colors import ListedColormap
```

```python
from sklearn.datasets import make_circles
x,y = make_circles(150,factor=0.1,noise=0.1)
plt.scatter(x[:,0],x[:,1],c=y,s=50,cmap='bwr',alpha=0.5)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=41)
classifier = SVC(kernel="linear")
classifier.fit(x_train,y_train.ravel())  # .ravel() is used to flatten an array into a 1 dimensional array
```

```python
from sklearn.metrics import accuracy_score
y_pred = classifier.predict(x_test)
accuracy_score(y_test,y_pred)
```

too bad , not even random guessing

```python
from mlxtend.plotting import plot_decision_regions

plt.figure(figsize=(8, 6))
plot_decision_regions(X=x_train, y=y_train, clf=classifier)
plt.title('SVM Decision Boundary (Training Set)')
plt.xlabel('x1')
plt.ylabel('x2')
plt.show()
```

```python
def plot_3d(x,y):
  r = np.exp(-(x**2).sum(1))
  ax = plt.subplot(projection='3d')
  ax.scatter3D(x[:,0],x[:,1],r,c=y,s=100,cmap='bwr',alpha=0.5)
  ax.set_xlabel('x1')
  ax.set_ylabel('x2')
  ax.set_zlabel('y')
  return ax
```

how that 2D data got converted to 3D  :


how is 2D data converted to 3D :    
we use a function  
$r = e^{-d^2}$

where $d^2 = x_1^2 + x_2^2$
it is also called as Radial Basis function or RBF

what it does , for points close to the center , these points are pushed high up and points which are further from the center , these remain low near the bottom



```python
plot_3d(x,y)
```

```python
rbf_classifier = SVC(kernel="rbf")
rbf_classifier.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
y_pred = rbf_classifier.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
from mlxtend.plotting import plot_decision_regions

plt.figure(figsize=(8, 6))
plot_decision_regions(X=x_train, y=y_train, clf=rbf_classifier)
plt.title('RBF SVM Decision Boundary (Training Set)')
plt.xlabel('x1')
plt.ylabel('x2')
plt.show()
```
