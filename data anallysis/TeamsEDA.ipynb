```python
for i in range(2008,2016):
    cskwon=len(df[(df.winner=="Chennai Super Kings") & (df.season==i)])
    cskplayed = len(df[((df.team1=="Chennai Super Kings") | (df.team2=="Chennai Super Kings"))& (df.season==i)])
    cskwr = cskwon/cskplayed
    print(f"{i} : played : {cskplayed} , win : {cskwon} and winrate : {cskwr}")
```

```python
team_win_by_runs = df.groupby(by='winner').sum()['win_by_runs']
team_win_by_runs
```

```python
team_win_by_runs = team_win_by_runs.drop(['Gujarat Lions' , 'Rising Pune Supergiants' , 'Kochi Tuskers Kerala' , 'Deccan Chargers' , 'Pune Warriors'])
team_win_by_runs
```

```python
team_win_by_runs.plot(kind = 'bar',title = 'win by runs')
```

```python
team_win_by_wickets = df.groupby(by='winner').sum()['win_by_wickets']
team_win_by_wickets
```

```python
team_win_by_wickets = team_win_by_wickets.drop(['Gujarat Lions' , 'Rising Pune Supergiants' , 'Kochi Tuskers Kerala' , 'Deccan Chargers' , 'Pune Warriors'])
team_win_by_wickets
```

```python
team_win_by_wickets.plot(kind = 'bar',title = 'win by wickets' , color = 'orange')
```

```python
overall_winners = pd.concat([team_win_by_runs, team_win_by_wickets], axis=1)
overall_winners.columns = ['win_by_runs', 'win_by_wickets']
overall_winners
```

```python
overall_winners.plot(kind = 'bar' , title = 'overall winners')
```
