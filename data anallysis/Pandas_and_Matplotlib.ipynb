```python
# grouping

dept = pd.Series(["IT","HR","IT","HR","Finance","Marketing","Sales","IT","Finance","HR","Marketing","Sales","IT","Finance","HR"])

salary = pd.Series([50000,40000,60000,45000,55000,48000,52000,65000,58000,47000,50000,54000,70000,62000,46000])

df = pd.DataFrame({"Department" : dept , "Salary" : salary})
print("\n")
grouped_mean = df.groupby("Department")["Salary"].mean()
grouped_median = df.groupby("Department")["Salary"].median()
print(f"{grouped_mean} \n")
print(f"{grouped_median} \n")
```

```python

# cleaning null values

df["Age"] = df["Age"].fillna(df["Age"].mean())  # fill mean values for ages
df.dropna(inplace = True)
print(df)
```

```python
# data filtering and querying

filtered = df[df["Salary"] > 60000]
print(filtered)
print("\n")
filtered_query = df.query("Age > 30")
print(filtered_query)
```

```python
# data manipulation and transform

df["Bonus"] = df["Salary"].apply(lambda x : x / 10)
print(df[["Name" , "Bonus" , "Salary"]])
```

```python
# group data

grouped = df.groupby("Department")["Salary"].mean()
print(grouped)
```

```python
# Identify and display duplicate rows

duplicate_rows = df[df.duplicated()]
print("Duplicate Rows: \n")
print(duplicate_rows)
```

```python
import matplotlib.pyplot as plt
```

```python
ipl_df.head()
```

```python

# line plot

plt.figure(figsize = (12,6))
plt.plot(ipl_df['team1_runs'].tail(100))
plt.xlabel("match index")
plt.ylabel("runs")
plt.title("runs trend (last 100 matches)")
plt.show()
```

```python
# multi line plot

plt.figure(figsize = (12,6))
plt.plot(ipl_df['team1_runs'].tail(100) , label = 'team runs' , color = 'purple')
plt.plot(ipl_df['team1_wickets'].tail(100) , label = 'team wickets' , color = 'orange')
plt.title("Runs vs Wickets (last 100 matches)")
plt.legend()
plt.grid(True)
plt.show()
```

```python

plt.figure(figsize = (12,6))
plt.plot(ipl_df['team1_runs'].tail(100) , label = 'team1_runs' , color = 'purple')
plt.plot(ipl_df['team2_runs'].tail(100) , label = 'team2_runs' , color = 'orange')
plt.title("team1_runs vs team2_runs (last 100 matches)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# scatter plot

match_indices = ipl_df.index[-100:] # last 100 matches
plt.scatter(match_indices , ipl_df['team1_runs'].tail(100) , c = 'blue' , label = 'team1_runs')
plt.scatter(match_indices , ipl_df['team2_runs'].tail(100) , c = 'red' , label = 'team2_runs')
plt.xlabel('match Index')
plt.ylabel('runs')
plt.title("team 1 and team 2 runs (in last 100 matches)")
plt.legend()
plt.show()
```

```python
# histogram

plt.hist(ipl_df['team2_wickets'].tail(100) , bins = 10)
plt.xlabel('wickets')
plt.ylabel('number of matches')
plt.title('distribution of team 2 wickets')
plt.show()
```

```python
# bar chart

toss_decisions = ipl_df['toss_decision'].value_counts()
plt.figure(figsize= (8,6))
plt.bar(toss_decisions.index ,  toss_decisions.values , color = ['skyblue' , 'lightcoral'])
plt.xlabel('toss decision')
plt.ylabel('number of matches')
plt.title('distribution of toss decisions')
plt.show()
```

```python
team_wins = ipl_df['winner'].value_counts()

plt.figure(figsize=(8,6))
plt.bar(team_wins.index , team_wins.values , color = ['skyblue' , 'lightcoral'])
plt.xlabel('winning teams')
plt.ylabel('number of matches')
plt.title('distribution of winning teams')
plt.xticks(rotation = 90)
plt.show()
```

```python
# box plot


plt.boxplot(ipl_df['team2_runs'].dropna() , orientation = 'horizontal')
plt.ylabel('team 2 runs')
plt.title('team 2 runs distribution')
plt.show()
```

```python
# pie chart

result = ipl_df['winner'].value_counts();
plt.figure(figsize=(8,8))
plt.pie(result.values , labels = result.index , pctdistance=1 , radius = 1)
plt.title('distribution of match result')
plt.xticks(rotation = 90)
plt.show()
```

```python
wicket_data = {
    '0-3' : len(ipl_df[ipl_df['team1_wickets'] <= 3]),
    '4-7' : len(ipl_df[(ipl_df['team1_wickets'] >= 4) & (ipl_df['team1_wickets'] <= 7)]),
    '8+' : len(ipl_df[ipl_df['team1_wickets'] >= 8])
}

plt.pie(wicket_data.values() , labels = wicket_data.keys() , autopct = '%1.1f%%')
plt.show()
```
